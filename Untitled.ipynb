##### We can reliably use 200 as our safe cut out value

### Data Cleaning

```python
taxi_data_filtered = taxi_data[(taxi_data['total_amount'] >= 0) & (taxi_data['total_amount'] < 200)]
```

```python
print(taxi_data.shape)
print(taxi_data_filtered.shape)
```
